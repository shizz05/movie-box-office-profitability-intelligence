# Movie Box Office & Profitability Intelligence
## Phase 2 – Data Cleaning
This notebook documents and validates the cleaning pipeline. Original raw CSV files are preserved.

```
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
movies = pd.read_csv(ROOT/'data/raw/tmdb_5000_movies.csv')
credits = pd.read_csv(ROOT/'data/raw/tmdb_5000_credits.csv')
print('Movies:', movies.shape, '| Credits:', credits.shape)
print('Duplicate movie IDs:', movies.id.duplicated().sum())
print('Duplicate credit IDs:', credits.movie_id.duplicated().sum())
```

## Financial eligibility
The project calculates profit and ROI only when both source budget and revenue are positive. Zero values are preserved in the source-derived master file.

```
eligible = (movies.budget > 0) & (movies.revenue > 0)
print('Eligible records:', eligible.sum())
print('Not eligible:', (~eligible).sum())
```
