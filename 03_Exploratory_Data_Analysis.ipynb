# Movie Box Office & Profitability Intelligence
## Phase 3 – Exploratory Data Analysis

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
df=pd.read_csv(ROOT/'data/cleaned/movies_cleaned.csv')
fin=df[(df.budget>0)&(df.revenue>0)].copy()
fin['profit']=fin.revenue-fin.budget
fin['roi_percentage']=fin.profit/fin.budget*100
print('Movies:',len(df),'| Financially eligible:',len(fin))
```

## Financial summary

```
display(fin[['budget','revenue','profit','roi_percentage']].describe().T)
print('Weighted aggregate ROI:', fin.profit.sum()/fin.budget.sum()*100)
```

## Budget vs revenue

```
plt.figure(figsize=(8,6)); plt.scatter(fin.budget/1e6,fin.revenue/1e6,alpha=.4,s=14)
plt.xlabel('Budget (USD millions)'); plt.ylabel('Revenue (USD millions)'); plt.title('Budget vs Revenue'); plt.show()
```

## Interpretation cautions
Profit and ROI are simplified budget-based measures. Multi-genre and multi-company associations can double-count movies if category totals are summed. Correlation does not establish causation.
